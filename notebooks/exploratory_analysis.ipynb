# E-Commerce Data Pipeline: Exploratory Analysis

This interactive notebook demonstrates how to inspect and query the synthetic e-commerce datasets generated by `src.data_generator`.

```python
import sys
from pathlib import Path
import pandas as pd

# Ensure project root is in sys.path
project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.config import CUSTOMERS_FILE, PRODUCTS_FILE, ORDERS_FILE, ORDER_ITEMS_FILE
```

## 1. Inspect Customers Dataset

```python
customers_df = pd.read_csv(CUSTOMERS_FILE)
print(f"Customers shape: {customers_df.shape}")
customers_df.head()
```

## 2. Inspect Products Catalog

```python
products_df = pd.read_csv(PRODUCTS_FILE)
print(f"Products shape: {products_df.shape}")
products_df.groupby('category').agg({'price': ['count', 'mean', 'min', 'max']})
```

## 3. Inspect Orders and Status Distribution

```python
orders_df = pd.read_csv(ORDERS_FILE)
print(f"Orders shape: {orders_df.shape}")
orders_df['order_status'].value_counts(normalize=True).mul(100).round(2)
```

## 4. Inspect Order Line Items

```python
items_df = pd.read_csv(ORDER_ITEMS_FILE)
print(f"Order items shape: {items_df.shape}")
items_df.describe()
```
